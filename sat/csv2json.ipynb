```python
from collections import defaultdict
from csv import reader
from datetime import timedelta
results = defaultdict(lambda: defaultdict(defaultdict))
t = 0
with open("res3.csv", "r") as res:
    rea = reader(res)
    next(rea)
    for expr_id, category, expr, opt, outcome, n_valid, n_vars, n_clauses, n_qubits, n_ancillas, mcx_count, depth, build_time, n_attempts, total_sim_time, total_iterations, wall_time in rea:
        t += float(wall_time)
        results[category][expr_id][opt] = {
            "outcome": outcome, 
            "n_valid": int(n_valid or 0), 
            "n_vars": int(n_vars or 0), 
            "n_clauses": int(n_clauses or 0), 
            "n_qubits": int(n_qubits or 0), 
            "n_ancillas": int(n_ancillas or 0), 
            "mcx_count": int(mcx_count or 0), 
            "depth": int(depth or 0), 
            "build_time": float(build_time or 0), 
            "n_attempts": int(n_attempts or 0), 
            "total_sim_time": float(total_sim_time or 0), 
            "total_iterations": int(total_iterations or 0), 
            "wall_time": float(wall_time or 0),
            "preprocess": float(wall_time or 0) - float(total_sim_time or 0) - float(build_time or 0)}
```

```python
from json import dump

with open("res3.json", "w") as re:
  dump(results, re)
```
